```python
import firebase_admin
import json
from firebase_admin import credentials, db

# Check if Firebase app is already initialized
if not firebase_admin._apps:
    cred = credentials.Certificate(r"coffeeshop-cb-firebase-adminsdk-fbsvc-6b988ec3e3.json")
    firebase_admin.initialize_app(cred, {
        "databaseURL": r"https://coffeeshop-cb-default-rtdb.asia-southeast1.firebasedatabase.app/"
    })


```

```python
# Reference to "items_info" node in Firebase
ref = db.reference("items_info")

# Read and upload JSONL data
with open(r"about_cs\product_info.jsonl", "r") as file:
    for line in file:
        product = json.loads(line.strip())  # Convert JSONL line to dictionary
        ref.push(product)  # Push each product as a new entry

print("All products uploaded successfully!")
```
