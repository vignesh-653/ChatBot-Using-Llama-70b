```python
import pandas as pd

# Load transaction data
transactions = pd.read_csv(r"cs_data\201904 sales reciepts.csv", usecols=["customer_id", "transaction_id", "transaction_date", "transaction_time", "product_id", "quantity", "unit_price"])

# Load product data
products = pd.read_csv(r"cs_data\product.csv", usecols=["product_id", "product", "product_category", "product_type"])

# Merge on product_id
df = transactions.merge(products, on="product_id", how="left")
```

```python
df['product'].unique()
```

```python
import re

# Function to extract size and clean product name
def clean_product_name(product):
    size_match = re.search(r"\b(Sm|Lg|Rg)\b", product)  # Find size
    size = size_match.group(0) if size_match else "Regular"  # Default to Regular if no size found
    product_name = re.sub(r"\b(Sm|Lg|Rg)\b", "", product).strip()  # Remove size from name
    return product_name, size

# Apply function to split name and size
df[["product", "size"]] = df["product"].apply(lambda x: pd.Series(clean_product_name(x)))

# Drop duplicates if any
df = df.drop_duplicates()

# Save the cleaned and merged data
df.to_csv("data.csv", index=False)

print("Data Cleaned")
```

```python
df.head()
```

```python
remove_categories = ["Coffee beans", "Loose Tea", "Branded"]

# Filter out these categories
df = df[~df["product_category"].isin(remove_categories)]

# Reset index after filtering
df = df.reset_index(drop=True)
# Save the cleaned and merged data
df.to_csv("data.csv", index=False)
# Display the cleaned dataframe
```

```python
df["day_of_week"] = pd.to_datetime(df["transaction_date"]).dt.dayofweek  # Monday=0, Sunday=6
df.drop(columns=["transaction_date"], inplace=True)  # Remove full date
```

```python
df.head()
```

```python
# Count unique transaction IDs per customer
customer_transaction_counts = df.groupby("customer_id")["transaction_id"].nunique()

# Keep customers who made more than one transaction
df = df[df["customer_id"].isin(customer_transaction_counts[customer_transaction_counts > 1].index)]
```

```python
import pandas as pd
from mlxtend.frequent_patterns import fpgrowth, association_rules

# Load dataset
df = pd.read_csv("cleaned_data.csv")  # Replace with actual file

# Convert transaction data into a basket format (one-hot encoded)
basket = df.groupby(['transaction_id', 'product'])['quantity'].sum().unstack().fillna(0)
basket = basket.astype(bool)  # Convert to Boolean (True/False)


```

```python
basket.shape
```

```python
print(basket.dtypes)  # Are all columns boolean?
print(basket.head())  # Confirm structure
```

```python
basket_small = basket.iloc[:, :]  # Use only the first 20 products
frequent_itemsets = fpgrowth(basket_small, min_support=0.04, use_colnames=True)
print(frequent_itemsets.shape)
```

```python
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1)
```

```python
rules.head()
```

```python
def recommend_items(cart_items, rules, top_n=5, alpha=0.7):
    """
    Recommend top N items based on FP-Growth rules with a balanced confidence & lift score.
    """
    cart_items = frozenset(cart_items)
    
    # Find rules where antecedents match cart items
    matching_rules = rules[rules['antecedents'].map(lambda x: cart_items & x == x)]

    if matching_rules.empty:
        return []
    
    # Compute a weighted score
    matching_rules['score'] = alpha * matching_rules['confidence'] + (1 - alpha) * matching_rules['lift']
    
    # Sort by score
    matching_rules = matching_rules.sort_values(by='score', ascending=False)
    
    # Extract top recommended items
    recommended_items = []
    seen_items = set()
    
    for _, rule in matching_rules.iterrows():
        for item in rule['consequents']:
            if item not in cart_items and item not in seen_items:
                recommended_items.append(item)
                seen_items.add(item)
            if len(recommended_items) == top_n:
                return recommended_items  # Stop at top N

    return recommended_items  # If less than N, return whatever is found


```
