```python
import requests
import time

long_query = """ 
 "role": "system",
                "content": (
                    "You are a friendly coffee shop assistant. Always reply as a JSON with two keys:\n"
                    "1. 'response': A helpful, concise reply to the user.\n"
                    "2. 'entities': A list of coffee or menu items mentioned (e.g., 'latte', 'croissant').\n"
                    "Only extract specific drink or food items, not names or locations.\n"
                    "Example format:\n"
                    "{\n"
                    '  "response": "Our Jamaican Coffee River is a popular choice. Would you like it black or with cream?",\n'
                    '  "entities": ["Jamaican Coffee River"]\n'
                    "}"


user: i want cappucnno
"""
start_time = time.time()  # Start the timer

response = requests.post("http://127.0.0.1:8004/chat",
                          json={"user_query": long_query},
                          stream=True)  # Enable streaming

first_token_time = None  # Track time for the first token

for chunk in response.iter_content(chunk_size=1024):
    if first_token_time is None:
        first_token_time = time.time()  # Time when the first token is received
    
    print(chunk.decode(), end="", flush=True)  # Print each chunk as received

end_time = time.time()  # End the timer

# Print timing results
if first_token_time:
    print(f"\nTime to first token: {first_token_time - start_time:.4f} seconds")
print(f"Total time taken: {end_time - start_time:.4f} seconds")
```
